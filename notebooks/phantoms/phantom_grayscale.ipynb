```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# ===================== Kích thước =====================
N = 160

# ===================== Mức hoạt độ chuẩn hóa =====================
activity = np.array([1.00, 0.75, 0.50, 0.25, 0.00])

# ===================== Tạo phantom =====================
phantom = np.zeros((N, N), dtype=np.float32)

phantom[:, 0:32]    = activity[0]
phantom[:, 32:64]   = activity[1]
phantom[:, 64:96]   = activity[2]
phantom[:, 96:128]  = activity[3]
phantom[:, 128:160] = activity[4]

# ===================== Hiển thị =====================
plt.figure(figsize=(5, 5), frameon=False)
plt.imshow(phantom,cmap="gray",vmin=0,vmax=1,interpolation="nearest")
plt.axis("off")
plt.subplots_adjust(left=0, right=1, top=1, bottom=0)
plt.show()

# ===================== Chuyển sang ảnh 8-bit =====================
phantom_8bit = np.round(phantom * 255).astype(np.uint8)

# ===================== Lưu ảnh 160 × 160 =====================
Image.fromarray(phantom_8bit,mode="L").save("phantommucxam.png")
```
